```python
    from pyspark.sql.functions import *
```

```python
file_path = "/Volumes/pyspark/spark_practice/csv_files/employees_data.csv"
```

```python
emp_df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(file_path)
emp_df.display()   
```

```python
#count of data
emp_df.select(count("*")).alias("total count").show()
```

```python
#particular dept count

emp_df.select(count("department")).show()

emp_df.groupBy(col("department")).agg(count("*")).alias("department count").display()
```

```python
#distinct count of departments

emp_df.select(countDistinct("department").alias("department count")).display()
```

```python
#total sum of salary

emp_df.select(sum(col("salary")).alias("sum of salary")).display()

#dept wise total salary

emp_df.groupBy(col("department")).agg(sum(col("salary")).alias("sum of department salary")).display()

#dept wise avg salary

emp_df.groupBy(col("department")).agg(avg(col("salary")).alias("avg salary of dept")).display()
```

```python
# min and max salary

emp_df.groupBy(col("department")).agg(max(col("salary")).alias("highest salary")).display()
emp_df.groupBy(col("department")).agg(min(col("salary")).alias("lowest salary")).display()
```

```python
emp_df.groupBy(col("department")).agg(max(col("salary")).alias("higest salary"), min(col("salary")).alias("lowest salary")).display()
```

```python
#first value/name display

emp_df.groupBy(col("department")).agg(first(col("first_name")).alias("first name")).display()
```

```python
#collect list of data

emp_df.groupBy(col("department")).agg(collect_list(col("first_name")).alias("first names")).display()

#collect set it will give distinct values

emp_df.groupBy(col("department")).agg(collect_set(col("salary")).alias("distincet salaries")).display()
```

```python
#median salary of 50th percentile

emp_df.select(percentile_approx(col("salary"),0.5)).display()
```

```python
#conditional counting

emp_df.groupBy(col("department")).agg(count("*").alias("total count"), sum(when(col("salary") > 80000,1).otherwise(0)).alias("high salary"), 
                                      sum(when(col("performance_score") > 4,1).otherwise(0)).alias("high performance")).display()
```
